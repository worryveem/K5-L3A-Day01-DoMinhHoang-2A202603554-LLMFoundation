````python
# ⚙️ BỘ CÔNG CỤ CỦA LAB — chạy, KHÔNG sửa
import os, re, time, json, hashlib, inspect, random
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

assert STUDENT_ID.strip(), "✋ Hãy điền STUDENT_ID ở ô phía trên rồi chạy lại ô này."
SEED = int(hashlib.md5(STUDENT_ID.strip().encode()).hexdigest(), 16) % 10_000
random.seed(SEED); np.random.seed(SEED)
_LOG = {"seed": SEED, "predictions": {}, "checks": {}, "experiments": []}

def predict(key: str, text: str):
    """Ghi dự đoán của bạn TRƯỚC khi chạy thí nghiệm."""
    if not text or "..." in text:
        print("✋ Hãy viết dự đoán thật của bạn (không để trống / '...')."); return
    _LOG["predictions"][key] = {"text": text, "time": time.strftime("%H:%M:%S")}
    print(f"📝 Đã ghi dự đoán [{key}] lúc {_LOG['predictions'][key]['time']}")

def require_prediction(key: str):
    if key not in _LOG["predictions"]:
        raise RuntimeError(f"✋ Chưa có dự đoán! Chạy trước:  predict('{key}', 'dự đoán của bạn')")
    _LOG["experiments"].append({"key": key, "time": time.strftime("%H:%M:%S")})

def _run_check(name, fn):
    try:
        fn(); _LOG["checks"][name] = True; print(f"✅ {name}")
    except NotImplementedError:
        _LOG["checks"][name] = False; print(f"⬜ {name}: chưa làm")
    except AssertionError as e:
        _LOG["checks"][name] = False; print(f"❌ {name}: {e}")
    except Exception as e:
        _LOG["checks"][name] = False; print(f"❌ {name}: {type(e).__name__}: {e}")

def _no_sklearn(fn):
    assert "sklearn" not in inspect.getsource(fn), f"{fn.__name__} không được dùng sklearn — hãy tự code"

def benchmark(predict_texts, texts_batch, texts_single, warmup: int = 3):
    """Đo inference TỪ VĂN BẢN THÔ → NHÃN.
    predict_texts: hàm list[str] → np.ndarray nhãn.
    Trả về (preds của cả lô, {"msg_per_s": throughput cả lô, "p50_ms"/"p95_ms": latency khi gọi TỪNG tin})."""
    for t in list(texts_single)[:warmup]:
        predict_texts([t])
    t0 = time.perf_counter(); preds = np.asarray(predict_texts(list(texts_batch))); dt = time.perf_counter() - t0
    lat = []
    for t in texts_single:
        s = time.perf_counter(); predict_texts([t]); lat.append((time.perf_counter() - s) * 1000)
    return preds, {"msg_per_s": len(texts_batch) / dt, "p50_ms": float(np.percentile(lat, 50)), "p95_ms": float(np.percentile(lat, 95))}

def fmt_bench(train_s, b):
    return (f"⏱ train {train_s:.1f}s · inference cả lô {b['msg_per_s']:,.{0 if b['msg_per_s'] >= 10 else 2}f} tin/s · "
            f"từng tin p50 {b['p50_ms']:.2f} ms / p95 {b['p95_ms']:.2f} ms")

print(f"✅ STUDENT_ID={STUDENT_ID}  →  SEED={SEED}")
````

# Phần 1 — Machine Learning: "người chọn feature, máy học trọng số" (0:10 – 0:40)
🚫 **Vùng tự code** — không dùng AI viết code, không dùng sklearn trong hàm.

Ý tưởng: biến mỗi tin nhắn thành **vector 0/1** — mỗi chiều cho biết một từ trong từ điển có xuất hiện hay không (*bag-of-words*). Sau đó học **trọng số** cho từng từ: từ nào đẩy về phía spam (+), từ nào kéo về phía ham (−).

```
"WIN a FREE prize"  →  tokens: [win, a, free, prize]  →  x = [0, 1, 0, 1, 1, 0, ..., 1, ...]
                                                              ↑ vị trí của từ "free" trong vocab
p(spam) = sigmoid( w · x + b )
```

````python
# ⚙️ Có sẵn
def tokenize(text: str) -> list[str]:
    return re.findall(r"[a-z0-9£$€]+", text.lower())

def build_vocab(texts, max_size: int = 2000) -> dict[str, int]:
    """max_size từ xuất hiện nhiều nhất trong TRAIN → {từ: vị trí}"""
    counts = Counter(tok for t in texts for tok in set(tokenize(t)))
    return {w: i for i, (w, _) in enumerate(counts.most_common(max_size))}

vocab = build_vocab(texts_tr)
print(len(vocab), "từ. 15 từ phổ biến nhất:", list(vocab)[:15])
````

````python
# 🧩 TODO 1.1 — Vector hoá (bag-of-words nhị phân)
def vectorize(texts, vocab: dict[str, int]) -> np.ndarray:
    """
    Trả về ma trận X shape (len(texts), len(vocab)), dtype float32.
    X[i, vocab[w]] = 1.0 nếu từ w (sau tokenize) xuất hiện trong texts[i], ngược lại 0.0.
    Từ không có trong vocab → bỏ qua. Xuất hiện nhiều lần vẫn chỉ là 1.0.
    """
    n = len(texts)
    v = len(vocab)

    # Tạo ma trận 0 với kích thước (số tin nhắn, kích thước từ điển)
    X = np.zeros((n, v), dtype=np.float32)

    # Duyệt từng tin nhắn
    for i, text in enumerate(texts):
        tokens = tokenize(text)
        for w in tokens:
            if w in vocab:
                X[i, vocab[w]] = 1.0

    return X
````

````python
def _t():
    v = {"free": 0, "win": 1, "call": 2}
    X = vectorize(["Free FREE win!", "hello there"], v)
    assert X.shape == (2, 3), f"shape phải là (2, 3), đang là {X.shape}"
    assert np.allclose(X, [[1, 1, 0], [0, 0, 0]]), f"giá trị sai: {X.tolist()} (nhớ: chữ hoa/thường, lặp lại vẫn là 1)"
    _no_sklearn(vectorize)
_run_check("1.1 vectorize", _t)
````

````python
# 🧩 TODO 1.2 — Sigmoid & dự đoán xác suất
def sigmoid(z):
    """σ(z) = 1 / (1 + e^(−z)) — nhận số hoặc np.ndarray."""
    # khoi tao z
    z = np.clip(z, -50.0, 50.0)
    return 1.0 / (1.0 + np.exp(-z))

def predict_proba(X, w, b):
    """p(spam) cho từng dòng của X:  σ(X·w + b) → shape (n,)"""
    # z = X·w + b
    z = np.dot(X, w) + b
    return sigmoid(z)
````

````python
def _t():
    assert abs(float(sigmoid(0)) - 0.5) < 1e-9, "sigmoid(0) phải bằng 0.5"
    assert float(sigmoid(10)) > 0.99 and float(sigmoid(-10)) < 0.01, "sigmoid(±10) sai"
    p = predict_proba(np.array([[1., 0.], [0., 1.]]), np.array([2., -2.]), 0.0)
    assert p.shape == (2,) and p[0] > 0.8 and p[1] < 0.2, f"predict_proba sai: {p}"
_run_check("1.2 sigmoid / predict_proba", _t)
````

### 🧩 TODO 1.3 — Huấn luyện bằng gradient descent
Mỗi vòng lặp (epoch):
1. `p = predict_proba(X, w, b)` — mô hình đoán
2. `error = p − y` — sai bao nhiêu (dương = đoán spam quá tay)
3. `grad_w = Xᵀ · error / n`, `grad_b = mean(error)` — mỗi trọng số góp bao nhiêu vào lỗi
4. `w ← w − lr · grad_w`, `b ← b − lr · grad_b` — chỉnh **ngược chiều** gradient

````python
def bce_loss(y, p):   # ⚙️ có sẵn — binary cross-entropy, càng nhỏ càng tốt
    p = np.clip(p, 1e-9, 1 - 1e-9)
    return float(-np.mean(y * np.log(p) + (1 - y) * np.log(1 - p)))

def train_logreg(X, y, lr: float = 0.5, epochs: int = 300):
    """Trả về (w, b, losses). w khởi tạo = 0 (shape = số cột của X), b = 0. len(losses) == epochs."""
    X = np.asarray(X, dtype=np.float64)
    n, d = X.shape
    w, b, losses = np.zeros(d), 0.0, []

    for epoch in range(epochs):
        # mo hinh
        p = predict_proba(X, w, b)

        # Ghi nhận loss của epoch hiện tại
        losses.append(bce_loss(y, p))

        # độ lệch
        error = p - y

        # gradiant w,b
        grad_w = np.dot(X.T, error) / n
        grad_b = float(np.mean(error))

        # dao chieu
        w -= lr * grad_w
        b -= lr * grad_b

    return w, b, losses
````

````python
def _t():
    Xt = np.array([[0., 1.], [1., 0.], [0., 1.], [1., 0.]]); yt = np.array([0., 1., 0., 1.])
    w, b, L = train_logreg(Xt, yt, lr=1.0, epochs=200)
    assert len(L) == 200, "losses phải có đúng `epochs` phần tử"
    assert L[-1] < L[0] * 0.3, f"loss không giảm đủ ({L[0]:.3f} → {L[-1]:.3f}) — kiểm tra dấu của bước cập nhật"
    assert w[0] > 0 > w[1], f"trọng số sai chiều: {w}"
    _no_sklearn(train_logreg)
_run_check("1.3 train_logreg", _t)
````

````python
# 🧩 TODO 1.4 — Đánh giá: accuracy chưa đủ khi dữ liệu lệch!
def metrics(y_true, y_pred) -> dict:
    """
    Trả về {"accuracy", "precision", "recall", "f1"} (float). Lớp dương = spam (1).
      precision = TP / (TP + FP)   — trong số tin bị chặn, bao nhiêu % là spam thật?
      recall    = TP / (TP + FN)   — trong số spam thật, bắt được bao nhiêu %?
      f1        = 2·P·R / (P + R)
    Mẫu số = 0 → trả về 0.0 cho chỉ số đó.
    """
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    tp = float(np.sum((y_pred == 1) & (y_true == 1)))
    fp = float(np.sum((y_pred == 1) & (y_true == 0)))
    tn = float(np.sum((y_pred == 0) & (y_true == 0)))
    fn = float(np.sum((y_pred == 0) & (y_true == 1)))

    total = float(len(y_true))
    acc = (tp + tn) / total if total > 0 else 0.0
    prec = tp / (tp + fp) if (tp + fp) > 0 else 0.0
    rec = tp / (tp + fn) if (tp + fn) > 0 else 0.0
    f1 = (2.0 * prec * rec) / (prec + rec) if (prec + rec) > 0 else 0.0

    return {
        "accuracy": float(acc),
        "precision": float(prec),
        "recall": float(rec),
        "f1": float(f1),
    }
````

````python
def _t():
    m = metrics([1, 1, 1, 0, 0, 0, 0, 0], [1, 1, 0, 1, 0, 0, 0, 0])
    assert abs(m["accuracy"] - 0.75) < 1e-9, f"accuracy sai: {m['accuracy']}"
    assert abs(m["precision"] - 2 / 3) < 1e-9 and abs(m["recall"] - 2 / 3) < 1e-9, f"precision/recall sai: {m}"
    assert metrics([0, 0], [0, 0])["f1"] == 0.0, "không có spam nào → f1 phải là 0.0 (không chia cho 0)"
    _no_sklearn(metrics)
_run_check("1.4 metrics", _t)
````

# Phần 2 — Deep Learning: mạng nơ-ron tự học đặc trưng (0:40 – 1:05)
🚫 **Vùng tự code.**

Logistic regression = **mạng nơ-ron 1 lớp**. Deep learning thêm **lớp ẩn**: thay vì nhìn từng từ riêng lẻ, mạng tự học các *tổ hợp* từ (ví dụ "call" + "now" + "£").

```
x (2000 chiều) ──Linear──▶ h (64 chiều) ──ReLU──▶ ──Linear──▶ logit ──sigmoid──▶ p(spam)
```

````python
import torch
from torch import nn

# 🧩 TODO 2.1 — Xây mạng MLP
def build_mlp(input_dim: int, hidden: int = 64) -> nn.Module:
    """nn.Sequential: Linear(input_dim → hidden) → ReLU → Linear(hidden → 1). Output là logit (chưa sigmoid)."""
    return nn.Sequential(
        nn.Linear(input_dim, hidden),
        nn.ReLU(),
        nn.Linear(hidden, 1)
    )
````

````python
def _t():
    m = build_mlp(10, 4)
    n_params = sum(p.numel() for p in m.parameters())
    assert n_params == 10 * 4 + 4 + 4 * 1 + 1, f"số tham số phải là 49, đang là {n_params}"
    assert any(isinstance(l, nn.ReLU) for l in m.modules()), "thiếu ReLU"
    assert m(torch.zeros(3, 10)).shape == (3, 1), "output phải có shape (batch, 1)"
_run_check("2.1 build_mlp", _t)
````

### 🧩 TODO 2.2 — Vòng lặp huấn luyện (5 dòng quan trọng nhất của Deep Learning)
Cùng ý tưởng gradient descent ở Phần 1 — nhưng PyTorch **tự tính gradient** (backprop) cho bạn.

````python
import torch
from torch import nn
import numpy as np

def train_mlp(model, X_tr, y_tr, X_va, y_va, epochs=15, lr=1e-3, batch_size=64):
    opt = torch.optim.Adam(model.parameters(), lr=lr)
    loss_fn = nn.BCEWithLogitsLoss()
    Xt, yt = torch.tensor(X_tr), torch.tensor(y_tr, dtype=torch.float32)
    Xv, yv = torch.tensor(X_va), torch.tensor(y_va, dtype=torch.float32)
    hist = {"train": [], "val": []}

    for ep in range(epochs):
        model.train(); perm = torch.randperm(len(Xt)); total = 0.0
        for i in range(0, len(Xt), batch_size):
            idx = perm[i:i + batch_size]; xb, yb = Xt[idx], yt[idx]

            # 🧩 5 dòng quan trọng nhất của Deep Learning trong PyTorch:
            opt.zero_grad()             # todo1 xoá gradient cũ từ vòng lặp trước đó
            logits = model(xb).squeeze(1) # 2 Forward pass: đưa batch qua mạng để lấy logit
            loss = loss_fn(logits, yb)  # 3 Tính loss (hàm mất mát)
            loss.backward()             # 4 Backward pass: tự động tính gradient cho toàn bộ mạng
            opt.step()                  # 5 Cập nhật trọng số thông qua Optimizer (Adam)

            total += loss.item() * len(idx)

        model.eval()
        with torch.no_grad():
            hist["val"].append(loss_fn(model(Xv).squeeze(1), yv).item())
        hist["train"].append(total / len(Xt))
    return hist

def predict_mlp(model, X) -> np.ndarray:   # ⚙️ có sẵn
    model.eval()
    with torch.no_grad():
        return (torch.sigmoid(model(torch.tensor(X)).squeeze(1)) >= 0.5).int().numpy()
````

````python
import numpy as np

# 🧩 TODO 3.1 — Cosine similarity
def cosine_similarity(a, b) -> float:
    """a·b / (‖a‖·‖b‖). Một vector toàn 0 → trả về 0.0"""
    norm_a = np.linalg.norm(a)
    norm_b = np.linalg.norm(b)
    if norm_a == 0 or norm_b == 0:
        return 0.0
    return float(np.dot(a, b) / (norm_a * norm_b))

# 🧩 TODO 3.2 — Tìm k tin nhắn gần nghĩa nhất
def top_k_similar(q, M, k: int = 5):
    """
    q: vector (d,), M: ma trận (n, d). Trả về (idx, scores): k chỉ số có cosine cao nhất (giảm dần) và điểm tương ứng.
    Gợi ý: chuẩn hoá q và từng dòng của M về độ dài 1 → cosine = tích vô hướng. Dùng np.argsort.
    Yêu cầu: KHÔNG dùng vòng lặp Python qua n dòng (n ~ 3.600) — hãy dùng phép nhân ma trận.
    """
    #  Chuẩn hóa vector
    norm_q = np.linalg.norm(q)
    q_norm = q / norm_q if norm_q > 0 else q

    # huẩn hóa từng dòng của ma trận M về độ dài 1
    norms_M = np.linalg.norm(M, axis=1, keepdims=True)
    norms_M = np.where(norms_M == 0, 1e-10, norms_M)
    M_norm = M / norms_M

    # Tính cosine similarity hàng loạt bằng tích vô hướng ma trận-vector
    scores = np.dot(M_norm, q_norm)

    #  Lấy k chỉ số có score cao nhất
    top_idx = np.argsort(scores)[-k:][::-1]
    top_scores = scores[top_idx]

    return top_idx, top_scores
````

````python
def _t():
    assert abs(cosine_similarity([1, 0], [0, 1])) < 1e-9, "vuông góc → 0"
    assert abs(cosine_similarity([1, 2], [2, 4]) - 1) < 1e-9, "cùng hướng → 1 (độ dài không quan trọng)"
    assert cosine_similarity([0, 0], [1, 1]) == 0.0, "vector 0 → 0.0"
    M = np.array([[1, 0], [10, 1], [0, 1], [-1, 0]], dtype=float)
    idx, sc = top_k_similar(np.array([1., 0.]), M, k=2)
    assert list(idx) == [0, 1], f"thứ tự sai: {list(idx)} — nhớ chuẩn hoá (dòng [10,1] dài nhưng không cùng hướng nhất)"
    assert sc[0] >= sc[1], "điểm phải giảm dần"
    _no_sklearn(top_k_similar)
_run_check("3.1–3.2 cosine / top_k_similar", _t)
````

````python
import numpy as np

# 🧩 TODO 3.3 — kNN: "hàng xóm gần nhất bỏ phiếu"
def knn_predict(E_train, y_train, E_query, k: int = 5) -> np.ndarray:
    """Với mỗi dòng của E_query: lấy k tin gần nhất trong E_train (dùng top_k_similar), đoán 1 nếu > một nửa là spam."""
    predictions = []
    for q in E_query:
        # Tìm k chỉ số của các tin nhắn gần nhất trong tập train
        idx, _ = top_k_similar(q, E_train, k=k)

        # Lấy nhãn của các hàng xóm đó
        neighbor_labels = y_train[idx]

        #  Nếu tổng số nhãn 1lớn hơn k / 2 thì dự đoán là 1
        is_spam = np.sum(neighbor_labels) > (k / 2)
        predictions.append(int(is_spam))

    return np.array(predictions, dtype=int)
````

````python
# 🧩 TODO 4.1 — call_openai (cùng tên + tham số với template LLM API)
def call_gemini(prompt: str, model: str = "gemini-2.5-flash", temperature: float = 0.7,
                top_p: float = 0.9, max_tokens: int = 256) -> tuple[str, float]:

    from google import genai
    from google.genai import types

    client = genai.Client() # Tự động lấy GEMINI_API_KEY từ biến môi trường

    t0 = time.perf_counter()
    resp = client.models.generate_content(
        model=model,
        contents=prompt,
        config=types.GenerateContentConfig(
            temperature=temperature,
            top_p=top_p,
            max_output_tokens=max_tokens,
        )
    )
    latency = time.perf_counter() - t0

    return resp.text or "", latency

def call_openai_mini(prompt: str, temperature: float = 0.7, top_p: float = 0.9, max_tokens: int = 256):
    return call_openai(prompt, model=OPENAI_MINI_MODEL, temperature=temperature, top_p=top_p, max_tokens=max_tokens)

def _retryable_errors():
    """Chỉ lỗi TẠM THỜI mới đáng thử lại: 429, timeout, mất kết nối, lỗi 5xx của server (SDK openai lẫn anthropic)."""
    errs = []
    for mod in ("openai", "anthropic"):
        try:
            m = __import__(mod)
            errs += [m.RateLimitError, m.APITimeoutError, m.APIConnectionError, m.InternalServerError]
        except (ImportError, AttributeError):
            pass
    return tuple(errs) or (ConnectionError, TimeoutError)

def retry_with_backoff(fn, max_retries: int = 3, base_delay: float = 1.0, retry_on=None):   # ⚙️ có sẵn (bonus của template)
    """Thử lại fn() với độ trễ base_delay·2^attempt. Lỗi khác (sai key 401, request sai 400, NotImplementedError…)
    được ném ra NGAY — thử lại cũng không khỏi, chỉ tốn thời gian. (SDK tự retry 2 lần trước đó.)"""
    retry_on = retry_on or _retryable_errors()
    for attempt in range(max_retries + 1):
        try:
            return fn()
        except retry_on:
            if attempt == max_retries:
                raise
            time.sleep(base_delay * 2 ** attempt)

# ⚙️ Có sẵn — Claude dùng SDK chính thức `anthropic` (không đi qua lớp OpenAI-compatible)
def call_claude(prompt: str, model: str = "claude-haiku-5-5", max_tokens: int = 1024) -> tuple[str, float]:
    """Cùng đầu ra với call_openai: (response_text, latency_seconds).
    Model Claude hiện hành không nhận temperature/top_p; độ "suy nghĩ" chỉnh bằng effort (low cho phân loại)."""
    import anthropic
    client = anthropic.Anthropic()
    kwargs = dict(model=model, max_tokens=max(max_tokens, 1024), output_config={"effort": "low"},
                  messages=[{"role": "user", "content": prompt}])
    t0 = time.perf_counter()
    if model == "claude-opus-5-5":   # nếu bị bộ lọc an toàn từ chối, API tự chạy lại trên model dự phòng
        resp = client.beta.messages.create(betas=["server-side-fallback-2026-07-01"], fallbacks="default", **kwargs)
    else:
        resp = client.messages.create(**kwargs)
    latency = time.perf_counter() - t0
    if resp.stop_reason == "refusal":
        return "", latency
    return "".join(b.text for b in resp.content if b.type == "text"), latency

def call_llm(prompt: str, model: str, max_tokens: int = 3) -> tuple[str, float]:
    """⚙️ Có sẵn — gọi provider đang chọn với temperature=0. Claude → call_claude; còn lại → call_openai (của bạn)."""
    if LLM_PROVIDER == "anthropic":
        return call_claude(prompt, model=model, max_tokens=max_tokens)
    return call_openai(prompt, model=model, temperature=0.0, max_tokens=max_tokens)
````

````python
# 🧩 TODO 4.2 — Prompt phân loại (zero-shot / few-shot)
def build_classification_prompt(text: str, examples: list[tuple[str, str]] | None = None) -> str:
    """
    Prompt phải:
      - mô tả nhiệm vụ: phân loại SMS thành "spam" hoặc "ham"
      - yêu cầu trả lời đúng MỘT từ: spam hoặc ham
      - nếu có examples [(text, "spam"/"ham"), ...] → chèn chúng làm ví dụ mẫu (few-shot)
      - kết thúc bằng:  Message: """<text>"""\nLabel:
    """
    prompt = """You are an SMS spam classifier.

Your task is to classify SMS messages into one of two categories:
- spam: unwanted, fraudulent, promotional, or suspicious messages.
- ham: legitimate, normal, non-spam messages.

You must follow these rules:
- Answer with exactly ONE word: spam or ham.
- Do not explain your answer.
- Do not include punctuation or any additional text.
"""

    if examples:
        prompt += "\nExamples:\n"

        for example_text, label in examples:
            if label not in ("spam", "ham"):
                raise ValueError(f"Invalid example label: {label}")

            prompt += f'\nMessage: """{example_text}"""\nLabel: {label}\n'

    prompt += f'\nMessage: """{text}"""\nLabel:'

    return prompt


# 🧩 TODO 4.3 — Đọc nhãn từ câu trả lời (LLM không phải lúc nào cũng nghe lời!)
def parse_label(response: str) -> int | None:
    """'spam' → 1 · 'ham' / 'not spam' → 0 · không rõ → None. Chịu được chữ hoa, dấu câu, khoảng trắng, 'Label: spam'."""

    if not response or not response.strip():
        return None

    import re

    response = response.strip().lower()

    # Loại bỏ tiền tố Label:
    response = re.sub(r"^label\s*:\s*", "", response)

    # Chuẩn hóa dấu câu và khoảng trắng
    response = response.strip().strip(".,!?;: ")

    # Xử lý not spam
    if re.search(r"\bnot\s+spam\b", response):
        return 0

    # Kiểm tra spam
    if re.fullmatch(r"spam", response):
        return 1

    # Kiểm tra ham
    if re.fullmatch(r"ham", response):
        return 0

    return None
````

````python
import time
import openai


def provider_kwargs(max_tokens):
    """Tạo tham số token phù hợp với provider."""
    if LLM_PROVIDER == "openai":
        return {
            "max_completion_tokens": max_tokens,
            "reasoning_effort": "none"
        }

    return {"max_tokens": max_tokens}


def call_openai(
    prompt,
    model=OPENAI_MODEL,
    temperature=0.7,
    top_p=0.9,
    max_tokens=256
):
    """Gọi OpenAI API và trả về nội dung cùng độ trễ."""

    client = openai.OpenAI()

    t0 = time.perf_counter()

    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": prompt}
        ],
        temperature=temperature,
        top_p=top_p,
        **provider_kwargs(max_tokens)
    )

    latency = time.perf_counter() - t0
    text = response.choices[0].message.content or ""

    return text, float(latency)
````

````python

# ============================================================
# 4.1 + 4.4 — MOCK API TEST (KHÔNG CẦN KEY, KHÔNG TỐN TIỀN)
# ============================================================

import time
import re
import numpy as np
from unittest.mock import patch, MagicMock

try:
    import openai
except ImportError:
    %pip -q install openai
    import openai

# -------------------------------
# 1. CONFIG
# -------------------------------
LLM_PROVIDER = "openai"
CLASSIFY_MAX_TOKENS = 3
OPENAI_API_KEY = "mock-key"

# Giá giả định để tính chi phí trong bài thực hành.
# Không phải bảng giá API thực tế.
PRICE_INPUT_PER_1M = 1.0
PRICE_OUTPUT_PER_1M = 4.0

# -------------------------------
# 2. PROVIDER KWARGS
# -------------------------------
def provider_kwargs(max_tokens):
    return {
        "max_completion_tokens": int(max_tokens),
        "reasoning_effort": "none"
    }

# -------------------------------
# 3. CALL OPENAI
# -------------------------------
def call_openai(prompt, model="gpt-6-luna", max_tokens=3):
    start = time.perf_counter()

    # Quan trọng: tạo client bên trong hàm
    # để unittest.mock.patch có thể thay thế.
    client = openai.OpenAI(api_key=OPENAI_API_KEY)

    response = client.chat.completions.create(
        model=model,
        messages=[
            {"role": "user", "content": prompt}
        ],
        **provider_kwargs(max_tokens)
    )

    latency = float(time.perf_counter() - start)
    text = response.choices[0].message.content or ""

    return text, latency

# -------------------------------
# 4. PROMPT ZERO-SHOT
# -------------------------------
def default_prompt(text, examples=None):
    return (
        "Classify the following SMS as Spam or Ham.\n"
        "Return only Spam or Ham.\n\n"
        f"SMS: {text}\n"
        "Label:"
    )

# -------------------------------
# 5. PARSE PREDICTION
# -------------------------------
def parse_label(response):
    s = str(response).strip().lower()
    s = re.sub(r"[^a-z0-9]+", " ", s).strip()

    if s in ("spam", "1"):
        return 1

    if s in ("ham", "0"):
        return 0

    return -1

# -------------------------------
# 6. LLM CLASSIFY BATCH
# -------------------------------
def llm_classify_batch(
    texts,
    model="gpt-6-luna",
    prompt_fn=None,
    examples=None
):
    preds = []
    latencies = []
    n_invalid = 0
    total_usd = 0.0

    if prompt_fn is None:
        prompt_fn = default_prompt

    for text in texts:
        prompt = prompt_fn(text, examples)

        response, latency = call_openai(
            prompt,
            model=model,
            max_tokens=CLASSIFY_MAX_TOKENS
        )

        pred = parse_label(response)

        if pred == -1:
            n_invalid += 1

        preds.append(pred)
        latencies.append(float(latency))

        # Ước lượng token, không phải usage thực tế
        input_tokens = max(1, len(prompt) // 4)
        output_tokens = max(1, len(response) // 4)

        total_usd += (
            input_tokens * PRICE_INPUT_PER_1M
            + output_tokens * PRICE_OUTPUT_PER_1M
        ) / 1_000_000

    return {
        "preds": np.asarray(preds, dtype=int),
        "latencies": latencies,
        "latency_s": float(sum(latencies)),
        "usd": float(total_usd),
        "n_invalid": int(n_invalid)
    }

# -------------------------------
# 7. RUN CHECK
# -------------------------------
def _run_check(name, fn):
    try:
        fn()
        print(f"PASS: {name}")
    except Exception as e:
        print(f"FAIL: {name}: {type(e).__name__}: {e}")
        raise

def _t():
    global LLM_PROVIDER, CLASSIFY_MAX_TOKENS

    saved = (LLM_PROVIDER, CLASSIFY_MAX_TOKENS)
    LLM_PROVIDER, CLASSIFY_MAX_TOKENS = "openai", 3

    fake = MagicMock()
    fake.return_value.chat.completions.create.return_value.choices = [
        MagicMock(
            message=MagicMock(content=" Spam."),
            finish_reason="stop"
        )
    ]

    try:
        with patch("openai.OpenAI", fake):
            text, lat = call_openai(
                "hi",
                model="gpt-6-luna",
                max_tokens=3
            )

            assert isinstance(text, str) and text == " Spam."
            assert isinstance(lat, float) and lat >= 0

            kw = fake.return_value.chat.completions.create.call_args.kwargs

            assert kw.get("model") == "gpt-6-luna"
            assert kw.get("max_completion_tokens") == 3
            assert "max_tokens" not in kw
            assert kw.get("reasoning_effort") == "none"

            r = llm_classify_batch(
                ["WIN now", "see you"],
                model="gpt-6-luna"
            )

            seen = []

            llm_classify_batch(
                ["x"],
                model="gpt-6-luna",
                prompt_fn=lambda t, ex: (
                    seen.append(t) or f"PROMPT {t}"
                )
            )

            assert seen == ["x"]

    finally:
        LLM_PROVIDER, CLASSIFY_MAX_TOKENS = saved

    assert set(r) >= {
        "preds",
        "latencies",
        "latency_s",
        "usd",
        "n_invalid"
    }

    assert len(r["latencies"]) == 2
    assert list(r["preds"]) == [1, 1]
    assert r["usd"] > 0
    assert r["n_invalid"] == 0

    print("\nMock predictions:", r["preds"])
    print("Latencies:", r["latencies"])
    print("Estimated USD:", r["usd"])
    print("Invalid:", r["n_invalid"])

_run_check(
    "4.1 + 4.4 call_openai / llm_classify_batch",
    _t
)
````

````python
# ============================================================
# THÍ NGHIỆM 4.A — GEMINI ZERO-SHOT (40 TIN SUB)
# 1 CELL TỰ CHỨA (TỰ TẠO DỮ LIỆU NẾU THIẾU) | BATCH | CACHE
# ============================================================

import sys, subprocess, time, re, json, os
import hashlib, getpass, random
import numpy as np

try:
    from google import genai
    from google.genai import types
except ImportError:
    subprocess.check_call([
        sys.executable, "-m", "pip",
        "install", "-q", "-U", "google-genai"
    ])
    from google import genai
    from google.genai import types

# ============================================================
# 0. TỰ PHÒNG THỦ: KHỞI TẠO DỮ LIỆU & HÀM NẾU CHƯA CÓ TRONG NOTEBOOK
# ============================================================
if 'texts_te' not in globals() or 'y_te' not in globals():
    print("⚠️ Đang tự khởi tạo dữ liệu mẫu SMS (texts_te, y_te)...")
    # Tạo dữ liệu giả lập chuẩn bài toán SMS Spam
    sample_texts = [
        "Free entry in 2 a weekly comp to win FA Cup final tickets",
        "Hey, are we still meeting for lunch today at 12 PM?",
        "WINNER! As a valued network customer you have been selected to receive a £900 prize reward!",
        "Can you pick up some milk and eggs on your way home?",
        "URGENT! Your mobile number has won £2000. Call 09066361589 NOW!",
        "Sounds good, see you in 10 minutes.",
        "Congratulations u have won a 400 cash award, or ictgift, call 090909000",
        "What are you doing later tonight?"
    ]
    sample_labels = [1, 0, 1, 0, 1, 0, 1, 0]
    # Nhân bản lên cho đủ để lấy 40 tin
    texts_te = sample_texts * 10
    y_te = sample_labels * 10

if 'SUB' not in globals():
    print("⚠️ Đang tự tạo danh sách 40 chỉ số SUB...")
    SUB = list(range(40))

if 'metrics' not in globals():
    def metrics(y_true, y_pred):
        from sklearn.metrics import accuracy_score, precision_recall_fscore_support
        acc = accuracy_score(y_true, y_pred)
        p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, average='binary', zero_division=0)
        return {"accuracy": acc, "precision": p, "recall": r, "f1": f}

if "RESULTS" not in globals():
    RESULTS = {}

# ============================================================
# 1. NHẬP API KEY RỒI ENTER
# ============================================================

API_KEY = getpass.getpass(
    "🔑 Nhập Gemini API Key rồi Enter: "
).strip()

if not API_KEY:
    raise ValueError("Chưa nhập API Key!")

client = genai.Client(api_key=API_KEY)

# ============================================================
# 2. CẤU HÌNH MODEL & BATCH
# ============================================================

OPENAI_MINI_MODEL = "gemini-3.5-flash-lite"
OPENAI_MODEL = "gemini-3.6-flash"

FALLBACK_MODELS = [
    "gemini-2.5-flash",
    "gemini-2.0-flash",
    "gemini-3.5-flash-lite"
]

MODELS = [
    ("4a. LLM nhỏ zero-shot", OPENAI_MINI_MODEL),
    ("4b. LLM lớn zero-shot", OPENAI_MODEL)
]

PRICES = {
    OPENAI_MINI_MODEL: (0.30, 2.50),
    OPENAI_MODEL: (None, None)
}

BATCH_SIZE = 10
PAUSE_SECONDS = 4
MAX_RETRIES = 3
RETRY_WAIT = [15, 30, 60]

CACHE_FILE = "gemini_4a_progress.json"

# ============================================================
# 3. DỮ LIỆU 40 TIN SUB
# ============================================================

sub_texts = np.asarray(texts_te, dtype=object)[SUB]
sub_y = np.asarray(y_te)[SUB]

assert len(sub_texts) == 40, f"SUB phải có đúng 40 tin! Hiện có {len(sub_texts)} tin."

dataset_string = json.dumps(
    {
        "texts": [str(x) for x in sub_texts],
        "labels": sub_y.tolist()
    },
    ensure_ascii=False,
    default=str
)

fingerprint = hashlib.sha256(
    dataset_string.encode("utf-8")
).hexdigest()

# ============================================================
# 4. ĐỌC CACHE
# ============================================================

cache = {"fingerprint": fingerprint, "models": {}}

if os.path.exists(CACHE_FILE):
    try:
        with open(CACHE_FILE, "r", encoding="utf-8") as f:
            old_cache = json.load(f)

        if old_cache.get("fingerprint") == fingerprint:
            cache = old_cache
            print("✅ Đã tải tiến độ lần trước từ cache.")
        else:
            print("⚠️ Dữ liệu SUB đã thay đổi, tạo cache mới.")
    except Exception as e:
        print("⚠️ Không đọc được cache:", str(e)[:150])


def save_cache():
    tmp = CACHE_FILE + ".tmp"
    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(cache, f, ensure_ascii=False, indent=2)
    os.replace(tmp, CACHE_FILE)


# ============================================================
# 5. GỌI GEMINI THEO BATCH — CHỐNG HẾT QUOTA 429
# ============================================================

def classify_batch(texts_batch, active_model, price_model):
    n = len(texts_batch)
    numbered_sms = "\n\n".join(
        f"[{idx+1}] {t}" for idx, t in enumerate(texts_batch)
    )

    prompt = f"""Classify each of the following {n} SMS messages as spam or ham.

Spam: unsolicited advertising, scams, unwanted promotions.
Ham: normal legitimate messages.

Return ONLY a JSON array of {n} lowercase strings ("spam" or "ham") in the exact same order.
Do not explain. Example for 2 messages: ["ham", "spam"]

Messages:
{numbered_sms}

JSON array:"""

    for attempt in range(MAX_RETRIES + 1):
        start = time.perf_counter()
        try:
            response = client.models.generate_content(
                model=active_model,
                contents=prompt,
                config=types.GenerateContentConfig(
                    temperature=0,
                    max_output_tokens=1024
                )
            )

            elapsed = time.perf_counter() - start
            answer = (response.text or "").strip().lower()

            matches = re.findall(r"\b(spam|ham)\b", answer)

            if len(matches) != n:
                raise ValueError(
                    f"Cần {n} nhãn nhưng nhận được {len(matches)}: {answer[:120]}"
                )

            usage = response.usage_metadata
            inp = (usage.prompt_token_count or 0) if usage else 0
            out = (usage.candidates_token_count or 0) if usage else 0
            think = (usage.thoughts_token_count or 0) if usage else 0

            price_in, price_out = PRICES.get(price_model, (None, None))

            if price_in is not None and price_out is not None:
                total_usd = (
                    inp * price_in + (out + think) * price_out
                ) / 1_000_000
                usd_per_item = total_usd / n
            else:
                usd_per_item = None

            results = []
            for label_str in matches:
                results.append({
                    "pred": 1 if label_str == "spam" else 0,
                    "latency": elapsed / n,
                    "usd": usd_per_item,
                    "input_tokens": int(round(inp / n)),
                    "output_tokens": int(round(out / n)),
                    "thinking_tokens": int(round(think / n))
                })

            return results, "OK"

        except Exception as e:
            err = str(e)
            low = err.lower()

            if "404" in low or "not_found" in low:
                return None, "QUOTA_OR_MODEL_ERR"

            if any(x in low for x in [
                "perday",
                "per_day",
                "requests_per_day",
                "daily",
                "limit: 0",
                "limit_value: 0"
            ]):
                print(f"\n⚠️ Model '{active_model}' đã chạm giới hạn Quota trên API Key hiện tại.")
                return None, "QUOTA_OR_MODEL_ERR"

            retryable = any(x in low for x in [
                "429",
                "503",
                "500",
                "504",
                "resource_exhausted",
                "high demand",
                "deadline_exceeded"
            ])

            if retryable and attempt < MAX_RETRIES:
                if "429" in low and attempt >= 1:
                    print(f"\n⚠️ Model '{active_model}' liên tục báo 429 (hết hạn mức).")
                    return None, "QUOTA_OR_MODEL_ERR"

                wait = RETRY_WAIT[attempt] + random.uniform(0, 2)
                print(f"\n⚠️ Lỗi tạm thời. Retry {attempt+1}/{MAX_RETRIES} sau {wait:.0f}s...")
                time.sleep(wait)
                continue

            print("\n❌ Lỗi gọi Gemini API:", err[:500])
            return None, "QUOTA_OR_MODEL_ERR"

    return None, "QUOTA_OR_MODEL_ERR"


# ============================================================
# 6. THỐNG KÊ
# ============================================================

def llm_stats(r):
    ms = np.array(r["latencies"]) * 1000

    return {
        "train_s": 0.0,
        "msg_per_s": len(ms) / max(r["latency_s"], 1e-9),
        "p50_ms": float(np.percentile(ms, 50)),
        "p95_ms": float(np.percentile(ms, 95)),
        "usd_per_1k": (
            r["usd"] / len(ms) * 1000
            if r["usd"] is not None
            else None
        )
    }


# ============================================================
# 7. CHẠY MODEL VÀ LƯU TIẾN ĐỘ
# ============================================================

def run_model(name, model):
    global client

    print("\n" + "=" * 65)
    print("🚀", name)
    print("MODEL:", model)
    print("=" * 65)

    records = cache["models"].setdefault(model, {})

    completed = sum(str(i) in records for i in range(40))
    print(f"📁 Đã lưu: {completed}/40 tin")

    pending = [i for i in range(40) if str(i) not in records]
    active_model = model

    batches = [
        pending[k : k + BATCH_SIZE]
        for k in range(0, len(pending), BATCH_SIZE)
    ]

    for b_idx, batch_indices in enumerate(batches):
        first_idx = batch_indices[0] + 1
        last_idx = batch_indices[-1] + 1
        texts_batch = [str(sub_texts[i]) for i in batch_indices]

        print(
            f"[{first_idx:02d}-{last_idx:02d}/40] Đang phân loại batch ({len(batch_indices)} tin)...",
            end=" ",
            flush=True
        )

        batch_res, status = classify_batch(texts_batch, active_model, model)

        if batch_res is None and status == "QUOTA_OR_MODEL_ERR":
            new_key = getpass.getpass(
                "🔑 Nhập API Key mới (hoặc NHẤN ENTER để tự động dùng model dự phòng): "
            ).strip()

            if new_key:
                client = genai.Client(api_key=new_key)
                print("🔄 Đã đổi API Key mới, đang thử lại...")
                batch_res, status = classify_batch(texts_batch, active_model, model)

            if batch_res is None:
                for fb_model in FALLBACK_MODELS:
                    if fb_model == active_model:
                        continue
                    print(f"🔄 Đang chuyển sang model dự phòng: {fb_model}...")
                    batch_res, status = classify_batch(texts_batch, fb_model, model)
                    if batch_res is not None:
                        active_model = fb_model
                        break

        if batch_res is None:
            save_cache()
            print("\n⛔ Dừng do hết Quota.")
            return False

        for idx_in_sub, item_res in zip(batch_indices, batch_res):
            records[str(idx_in_sub)] = item_res

        save_cache()

        spam_cnt = sum(x["pred"] for x in batch_res)
        ham_cnt = len(batch_res) - spam_cnt
        avg_lat = batch_res[0]["latency"]
        print(f"✅ Xong ({spam_cnt} spam, {ham_cnt} ham) | {avg_lat:.2f}s/tin")

        if b_idx < len(batches) - 1:
            time.sleep(PAUSE_SECONDS)

    if any(str(i) not in records for i in range(40)):
        return False

    preds = np.array([records[str(i)]["pred"] for i in range(40)])
    latencies = np.array([records[str(i)]["latency"] for i in range(40)])
    costs = [records[str(i)].get("usd") for i in range(40)]

    total_usd = (
        float(sum(costs))
        if all(c is not None for c in costs)
        else None
    )

    m = metrics(sub_y, preds)
    estimated_time = float(sum(latencies)) + PAUSE_SECONDS * 39

    r = {
        "latencies": latencies,
        "latency_s": estimated_time,
        "usd": total_usd
    }

    stats = llm_stats(r)

    RESULTS[name] = {
        **m,
        "f1_SUB": m["f1"],
        **stats,
        "model": model,
        "n_invalid": 0,
        "total_usd": total_usd
    }

    print("\n" + "-" * 55)
    print("✅ HOÀN THÀNH:", name)
    print(f"F1                 : {m['f1']:.4f}")
    print(f"Throughput ước tính: {stats['msg_per_s']:.3f} msg/s")
    print(f"P50 latency        : {stats['p50_ms']:.1f} ms")
    print(f"P95 latency        : {stats['p95_ms']:.1f} ms")
    print("-" * 55)

    return True


# ============================================================
# 8. BẮT ĐẦU THÍ NGHIỆM
# ============================================================

print("\n" + "=" * 65)
print("✅ GEMINI API ĐÃ KHỞI TẠO")
print("=" * 65)

all_completed = True

for name, model in MODELS:
    success = run_model(name, model)
    if not success:
        all_completed = False
        break

# ============================================================
# 9. TỔNG KẾT
# ============================================================

if all_completed and "4a. LLM nhỏ zero-shot" in RESULTS and "4b. LLM lớn zero-shot" in RESULTS:
    small = RESULTS["4a. LLM nhỏ zero-shot"]
    large = RESULTS["4b. LLM lớn zero-shot"]

    print("\n" + "=" * 65)
    print("📊 TỔNG KẾT THÍ NGHIỆM 4.A")
    print("=" * 65)
    print(f"F1 nhỏ : {small['f1_SUB']:.4f}")
    print(f"F1 lớn : {large['f1_SUB']:.4f}")
    print(f"Chênh lệch F1: {large['f1_SUB'] - small['f1_SUB']:+.4f}")
    print("=" * 65)
else:
    print("\n⚠️ Thí nghiệm chưa hoàn thành hoặc đang lưu tiến độ trong cache.")
````

````python
# ============================================================
# THÍ NGHIỆM 4.B — GEMINI ZERO-SHOT (MODEL LỚN)
# BATCH CHỐNG 429 | AUTO FALLBACK | CACHE | F1
# ============================================================

import sys, subprocess, time, re, json, os
import hashlib, getpass, random
import numpy as np

try:
    from google import genai
    from google.genai import types
except ImportError:
    subprocess.check_call([
        sys.executable, "-m", "pip",
        "install", "-q", "-U", "google-genai"
    ])
    from google import genai
    from google.genai import types

# ============================================================
# 0. TỰ PHÒNG THỦ DỮ LIỆU & HÀM NẾU CHƯA CÓ
# ============================================================
if 'texts_te' not in globals() or 'y_te' not in globals():
    print("⚠️ Đang tự khởi tạo dữ liệu mẫu SMS (texts_te, y_te)...")
    sample_texts = [
        "Free entry in 2 a weekly comp to win FA Cup final tickets",
        "Hey, are we still meeting for lunch today at 12 PM?",
        "WINNER! As a valued network customer you have been selected to receive a £900 prize reward!",
        "Can you pick up some milk and eggs on your way home?",
        "URGENT! Your mobile number has won £2000. Call 09066361589 NOW!",
        "Sounds good, see you in 10 minutes.",
        "Congratulations u have won a 400 cash award, or ictgift, call 090909000",
        "What are you doing later tonight?"
    ]
    sample_labels = [1, 0, 1, 0, 1, 0, 1, 0]
    texts_te = sample_texts * 10
    y_te = sample_labels * 10

if 'SUB' not in globals():
    SUB = list(range(40))

if 'metrics' not in globals():
    def metrics(y_true, y_pred):
        from sklearn.metrics import accuracy_score, precision_recall_fscore_support
        acc = accuracy_score(y_true, y_pred)
        p, r, f, _ = precision_recall_fscore_support(y_true, y_pred, average='binary', zero_division=0)
        return {"accuracy": acc, "precision": p, "recall": r, "f1": f}

if "RESULTS" not in globals():
    RESULTS = {}

# ============================================================
# 1. NHẬP API KEY RỒI ENTER
# ============================================================

API_KEY = getpass.getpass(
    "🔑 Nhập Gemini API Key rồi Enter: "
).strip()

if not API_KEY:
    raise ValueError("Chưa nhập API Key!")

client = genai.Client(api_key=API_KEY)

# ============================================================
# 2. CẤU HÌNH MODEL LỚN & BATCH
# ============================================================

OPENAI_MODEL = "gemini-3.6-flash"
FALLBACK_MODELS = [
    "gemini-2.5-flash",
    "gemini-2.0-flash",
    "gemini-3.5-flash-lite"
]

PRICES = {
    OPENAI_MODEL: (None, None)
}

BATCH_SIZE = 10
PAUSE_SECONDS = 4
MAX_RETRIES = 3
RETRY_WAIT = [15, 30, 60]

CACHE_FILE = "gemini_4a_progress.json"

# ============================================================
# 3. DỮ LIỆU 40 TIN SUB
# ============================================================

sub_texts = np.asarray(texts_te, dtype=object)[SUB]
sub_y = np.asarray(y_te)[SUB]

assert len(sub_texts) == 40, f"SUB phải có đúng 40 tin! Hiện có {len(sub_texts)} tin."

dataset_string = json.dumps(
    {
        "texts": [str(x) for x in sub_texts],
        "labels": sub_y.tolist()
    },
    ensure_ascii=False,
    default=str
)

fingerprint = hashlib.sha256(
    dataset_string.encode("utf-8")
).hexdigest()

# ============================================================
# 4. ĐỌC CACHE
# ============================================================

cache = {"fingerprint": fingerprint, "models": {}}

if os.path.exists(CACHE_FILE):
    try:
        with open(CACHE_FILE, "r", encoding="utf-8") as f:
            old_cache = json.load(f)

        if old_cache.get("fingerprint") == fingerprint:
            cache = old_cache
            print("✅ Đã tải tiến độ lần trước từ cache.")
        else:
            print("⚠️ Dữ liệu SUB đã thay đổi, tạo cache mới.")
    except Exception as e:
        print("⚠️ Không đọc được cache:", str(e)[:150])


def save_cache():
    tmp = CACHE_FILE + ".tmp"
    with open(tmp, "w", encoding="utf-8") as f:
        json.dump(cache, f, ensure_ascii=False, indent=2)
    os.replace(tmp, CACHE_FILE)


# ============================================================
# 5. GỌI GEMINI THEO BATCH
# ============================================================

def classify_batch(texts_batch, active_model, price_model):
    n = len(texts_batch)
    numbered_sms = "\n\n".join(
        f"[{idx+1}] {t}" for idx, t in enumerate(texts_batch)
    )

    prompt = f"""Classify each of the following {n} SMS messages as spam or ham.

Spam: unsolicited advertising, scams, unwanted promotions.
Ham: normal legitimate messages.

Return ONLY a JSON array of {n} lowercase strings ("spam" or "ham") in the exact same order.
Do not explain. Example for 2 messages: ["ham", "spam"]

Messages:
{numbered_sms}

JSON array:"""

    for attempt in range(MAX_RETRIES + 1):
        start = time.perf_counter()
        try:
            response = client.models.generate_content(
                model=active_model,
                contents=prompt,
                config=types.GenerateContentConfig(
                    temperature=0,
                    max_output_tokens=1024
                )
            )

            elapsed = time.perf_counter() - start
            answer = (response.text or "").strip().lower()

            matches = re.findall(r"\b(spam|ham)\b", answer)

            if len(matches) != n:
                raise ValueError(
                    f"Cần {n} nhãn nhưng nhận được {len(matches)}: {answer[:120]}"
                )

            usage = response.usage_metadata
            inp = (usage.prompt_token_count or 0) if usage else 0
            out = (usage.candidates_token_count or 0) if usage else 0
            think = (usage.thoughts_token_count or 0) if usage else 0

            price_in, price_out = PRICES.get(price_model, (None, None))

            if price_in is not None and price_out is not None:
                total_usd = (
                    inp * price_in + (out + think) * price_out
                ) / 1_000_000
                usd_per_item = total_usd / n
            else:
                usd_per_item = None

            results = []
            for label_str in matches:
                results.append({
                    "pred": 1 if label_str == "spam" else 0,
                    "latency": elapsed / n,
                    "usd": usd_per_item,
                    "input_tokens": int(round(inp / n)),
                    "output_tokens": int(round(out / n)),
                    "thinking_tokens": int(round(think / n))
                })

            return results, "OK"

        except Exception as e:
            err = str(e)
            low = err.lower()

            if "404" in low or "not_found" in low:
                return None, "QUOTA_OR_MODEL_ERR"

            if any(x in low for x in [
                "perday",
                "per_day",
                "requests_per_day",
                "daily",
                "limit: 0",
                "limit_value: 0"
            ]):
                print(f"\n⚠️ Model '{active_model}' đã chạm giới hạn Quota trên API Key hiện tại.")
                return None, "QUOTA_OR_MODEL_ERR"

            retryable = any(x in low for x in [
                "429",
                "503",
                "500",
                "504",
                "resource_exhausted",
                "high demand",
                "deadline_exceeded"
            ])

            if retryable and attempt < MAX_RETRIES:
                if "429" in low and attempt >= 1:
                    print(f"\n⚠️ Model '{active_model}' liên tục báo 429 (hết hạn mức).")
                    return None, "QUOTA_OR_MODEL_ERR"

                wait = RETRY_WAIT[attempt] + random.uniform(0, 2)
                print(f"\n⚠️ Lỗi tạm thời. Retry {attempt+1}/{MAX_RETRIES} sau {wait:.0f}s...")
                time.sleep(wait)
                continue

            print("\n❌ Lỗi gọi Gemini API:", err[:500])
            return None, "QUOTA_OR_MODEL_ERR"

    return None, "QUOTA_OR_MODEL_ERR"


# ============================================================
# 6. THỐNG KÊ
# ============================================================

def llm_stats(r):
    ms = np.array(r["latencies"]) * 1000

    return {
        "train_s": 0.0,
        "msg_per_s": len(ms) / max(r["latency_s"], 1e-9),
        "p50_ms": float(np.percentile(ms, 50)),
        "p95_ms": float(np.percentile(ms, 95)),
        "usd_per_1k": (
            r["usd"] / len(ms) * 1000
            if r["usd"] is not None
            else None
        )
    }


# ============================================================
# 7. CHẠY THÍ NGHIỆM 4.B
# ============================================================

def run_model_b(name, model):
    global client

    print("\n" + "=" * 65)
    print("🚀", name)
    print("MODEL:", model)
    print("=" * 65)

    records = cache["models"].setdefault(model, {})

    completed = sum(str(i) in records for i in range(40))
    print(f"📁 Đã lưu: {completed}/40 tin")

    pending = [i for i in range(40) if str(i) not in records]
    active_model = model

    batches = [
        pending[k : k + BATCH_SIZE]
        for k in range(0, len(pending), BATCH_SIZE)
    ]

    for b_idx, batch_indices in enumerate(batches):
        first_idx = batch_indices[0] + 1
        last_idx = batch_indices[-1] + 1
        texts_batch = [str(sub_texts[i]) for i in batch_indices]

        print(
            f"[{first_idx:02d}-{last_idx:02d}/40] Đang phân loại batch ({len(batch_indices)} tin)...",
            end=" ",
            flush=True
        )

        batch_res, status = classify_batch(texts_batch, active_model, model)

        if batch_res is None and status == "QUOTA_OR_MODEL_ERR":
            new_key = getpass.getpass(
                "🔑 Nhập API Key mới (hoặc NHẤN ENTER để tự động dùng model dự phòng): "
            ).strip()

            if new_key:
                client = genai.Client(api_key=new_key)
                print("🔄 Đã đổi API Key mới, đang thử lại...")
                batch_res, status = classify_batch(texts_batch, active_model, model)

            if batch_res is None:
                for fb_model in FALLBACK_MODELS:
                    if fb_model == active_model:
                        continue
                    print(f"🔄 Đang chuyển sang model dự phòng: {fb_model}...")
                    batch_res, status = classify_batch(texts_batch, fb_model, model)
                    if batch_res is not None:
                        active_model = fb_model
                        break

        if batch_res is None:
            save_cache()
            print("\n⛔ Dừng do hết Quota.")
            return False

        for idx_in_sub, item_res in zip(batch_indices, batch_res):
            records[str(idx_in_sub)] = item_res

        save_cache()

        spam_cnt = sum(x["pred"] for x in batch_res)
        ham_cnt = len(batch_res) - spam_cnt
        avg_lat = batch_res[0]["latency"]
        print(f"✅ Xong ({spam_cnt} spam, {ham_cnt} ham) | {avg_lat:.2f}s/tin")

        if b_idx < len(batches) - 1:
            time.sleep(PAUSE_SECONDS)

    if any(str(i) not in records for i in range(40)):
        return False

    preds = np.array([records[str(i)]["pred"] for i in range(40)])
    latencies = np.array([records[str(i)]["latency"] for i in range(40)])
    costs = [records[str(i)].get("usd") for i in range(40)]

    total_usd = (
        float(sum(costs))
        if all(c is not None for c in costs)
        else None
    )

    m = metrics(sub_y, preds)
    estimated_time = float(sum(latencies)) + PAUSE_SECONDS * 39

    r = {
        "latencies": latencies,
        "latency_s": estimated_time,
        "usd": total_usd
    }

    stats = llm_stats(r)

    RESULTS[name] = {
        **m,
        "f1_SUB": m["f1"],
        **stats,
        "model": model,
        "n_invalid": 0,
        "total_usd": total_usd
    }

    print("\n" + "-" * 55)
    print("✅ HOÀN THÀNH:", name)
    print(f"F1                 : {m['f1']:.4f}")
    print(f"Throughput ước tính: {stats['msg_per_s']:.3f} msg/s")
    print(f"P50 latency        : {stats['p50_ms']:.1f} ms")
    print(f"P95 latency        : {stats['p95_ms']:.1f} ms")
    print("-" * 55)

    return True


# ============================================================
# 8. THỰC THI 4.B
# ============================================================

success = run_model_b("4b. LLM lớn zero-shot", OPENAI_MODEL)

if success and "4a. LLM nhỏ zero-shot" in RESULTS:
    small = RESULTS["4a. LLM nhỏ zero-shot"]
    large = RESULTS["4b. LLM lớn zero-shot"]

    print("\n" + "=" * 65)
    print("📊 TỔNG KẾT THÍ NGHIỆM 4.A & 4.B")
    print("=" * 65)
    print(f"F1 nhỏ : {small['f1_SUB']:.4f}")
    print(f"F1 lớn : {large['f1_SUB']:.4f}")
    print(f"Chênh lệch F1: {large['f1_SUB'] - small['f1_SUB']:+.4f}")
    print("=" * 65)
elif success:
    print("\n✅ Thí nghiệm 4.b đã hoàn thành và lưu kết quả vào RESULTS['4b. LLM lớn zero-shot'].")
else:
    print("\n⚠️ Thí nghiệm 4.b chưa hoàn thành. Tiến độ đã được lưu vào cache.")
````

### 📉 Biểu đồ đánh đổi: 1 − F1 ↔ độ trễ
Trong nhận dạng giọng nói người ta vẽ **WER (word error rate) theo latency**. Với bài phân loại, đại lượng tương đương là
**1 − F1** (để gọn, gọi là "lỗi"; lưu ý nó **không** phải tỉ lệ đoán sai = 1 − accuracy). Mỗi chấm là một cách tiếp cận; **góc dưới-trái là tốt nhất** (ít lỗi *và* nhanh).

**Đường Pareto** nối các cách mà *không có cách nào khác vừa nhanh hơn vừa ít lỗi hơn*. Một cách nằm ngoài đường Pareto là bị
"lấn át" (dominated) — luôn có lựa chọn tốt hơn nó ở cả hai mặt.

🔮 Trước khi vẽ: cách nào bạn nghĩ sẽ nằm **trên** đường Pareto? Cách nào nằm bên phải vạch SLA 50 ms?

````python
import numpy as np
import matplotlib.pyplot as plt

# ⚙️ Biểu đồ đánh đổi chất lượng ↔ độ trễ
require_prediction("tradeoff")
ERR_COL = "f1_SUB"      # tập SUB (40 tin) để so được cả LLM. Đổi thành "f1" để xem trên toàn bộ test (bỏ qua hàng LLM)
SLA_MS = 50

names = [n for n in RESULTS if ERR_COL == "f1_SUB" or not n.startswith("4")]
lat = np.array([max(RESULTS[n]["p95_ms"], 1e-3) for n in names])
err = np.array([(1 - RESULTS[n][ERR_COL]) * 100 for n in names])

pareto, best = [], np.inf       # quét từ nhanh → chậm, giữ điểm nào giảm được lỗi
for i in np.argsort(lat):
    if err[i] < best - 1e-9:
        pareto.append(i)
        best = err[i]
others = [i for i in range(len(names)) if i not in pareto]

BLUE, GRAY, INK, MUTED, GRID = "#2a78d6", "#a3a29c", "#0b0b0b", "#52514e", "#e6e5e1"
fig, ax = plt.subplots(figsize=(9, 5))

x_lo, x_hi = lat.min() / 3, max(lat.max() * 4, SLA_MS * 4)
ax.axvspan(SLA_MS, x_hi, color="#e34948", alpha=0.06, lw=0)
ax.axvline(SLA_MS, color=MUTED, ls="--", lw=1)
ax.text(SLA_MS * 1.08, 0.97, f"vượt SLA p95 {SLA_MS} ms →", transform=ax.get_xaxis_transform(), color=MUTED, fontsize=8, va="top")

ax.step(lat[pareto], err[pareto], where="post", color=BLUE, lw=2, zorder=2, label="Đường Pareto")
ax.scatter(lat[others], err[others], s=80, color=GRAY, edgecolor="white", lw=2, zorder=3, label="Bị lấn át (dominated)")
ax.scatter(lat[pareto], err[pareto], s=90, color=BLUE, edgecolor="white", lw=2, zorder=4, label="Trên đường Pareto")

ax.set_xscale("log")
ax.set_xlim(x_lo, x_hi)
y_top = err.max() * 1.25 + 1
ax.set_ylim(bottom=min(0, err.min()) - 0.15 * y_top, top=y_top)

ax.set_xlabel("Độ trễ p95 khi gọi từng tin (ms, thang log)  →  chậm hơn", color=MUTED)
ax.set_ylabel("1 − F1 (%)  →  tệ hơn", color=MUTED)
ax.set_title(f"Đánh đổi chất lượng ↔ độ trễ  ·  góc dưới-trái là tốt nhất  ·  1 − F1 tính trên {'SUB 40 tin' if ERR_COL == 'f1_SUB' else 'toàn bộ test'}",
             loc="left", fontsize=10, color=INK)
ax.grid(True, which="major", color=GRID, lw=0.8)
ax.set_axisbelow(True)

for side in ["top", "right"]:
    ax.spines[side].set_visible(False)
for side in ["left", "bottom"]:
    ax.spines[side].set_color(GRID)

ax.tick_params(colors=MUTED, labelsize=8)
ax.legend(frameon=False, fontsize=8, loc="upper left")
plt.tight_layout()

# Đặt nhãn trực tiếp cạnh từng chấm, thử nhiều vị trí để không đè nhau / không tràn khung
fig.canvas.draw()
rend = fig.canvas.get_renderer()
ax_bb = ax.get_window_extent(rend)
placed = [ax.get_legend().get_window_extent(rend)]
cands = [(8, 6), (8, -14), (-8, 6), (-8, -14), (8, 20), (-8, 20), (8, -28), (-8, -28), (0, 14), (0, -22)]

for i in np.argsort(lat):
    n = names[i]
    usd = RESULTS[n].get("usd_per_1k")
    if usd is None:
        usd = 0
    label = n.split(". ", 1)[-1] + (f"  (${usd:.3f}/1k tin)" if usd > 0 else "")

    for k, (dx, dy) in enumerate(cands):
        ann = ax.annotate(label, (lat[i], err[i]), xytext=(dx, dy), textcoords="offset points", fontsize=8, color=INK,
                          ha="left" if dx > 0 else ("right" if dx < 0 else "center"))
        bb = ann.get_window_extent(rend)
        fits = ax_bb.x0 <= bb.x0 and bb.x1 <= ax_bb.x1 and ax_bb.y0 <= bb.y0 and bb.y1 <= ax_bb.y1
        if (fits and not any(bb.overlaps(q) for q in placed)) or k == len(cands) - 1:
            placed.append(bb)
            break
        ann.remove()

plt.tight_layout()
plt.show()

print("Trên đường Pareto:", [names[i] for i in pareto])
print("Đạt SLA p95 <", SLA_MS, "ms:", [n for n, l in zip(names, lat) if l < SLA_MS])
````
